```python
import math
import numpy as np
import matplotlib as plt
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# Configuration
n_samples = 500  # Total points (250 per cluster)
cluster_std = 0.6  # Controls the spread/size of the circles

# Define distinct centers so they don't overlap
centers = [[2, 3], [7, 3]]

# Generate the dataset
X, _ = make_blobs(
    n_samples=n_samples, 
    centers=centers, 
    cluster_std=cluster_std, 
    random_state=42
)

# Plot the dataset
plt.figure(figsize=(8, 5))
plt.scatter(X[:,0], X[:,1], color='crimson', label='Cluster 1', alpha=0.7, edgecolors='k', s=35)

plt.title("Two Non-Overlapping Circular Clusters")
plt.xlabel("X coordinate")
plt.ylabel("Y coordinate")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.axis('equal')  # Keeps the aspect ratio 1:1 so circles look round
plt.show()
```

Convergence Criteria:- The max distance between the two consecutive centroid of a cluster is epsilon = 0.005

```python
C1=updated_C1
C2=updated_C2
shift1=10.0
shift2=10.0
itr=0

while not (shift1<0.005 and shift2<0.005):
    sum1=0.0
    sum2=0.0
    s1=np.zeros(500)
    s2=np.zeros(500)
    for i in np.arange(0,500):
        s1[i]=math.pow(C1[0]-X[i][0],2)+math.pow(C1[1]-X[i][1],2)
        s2[i]=math.pow(C2[0]-X[i][0],2)+math.pow(C2[1]-X[i][1],2)
        
        

    d1=math.sqrt(np.sum(s1))
    d2=math.sqrt(np.sum(s2))


    X1=[]
    X2=[]

    for i in np.arange(0,500):
        if s1[i]<s2[i]:
            X1.append(X[i])
        else:
            X2.append(X[i])
    x1=0.0
    x2=0.0
    s1=0.0
    s2=0.0
    for j in np.arange(0,len(X1)):
        s1=X1[j][0]
        x1=x1+s1
    for j in np.arange(0,len(X1)):
        s2=X1[j][1]
        x2=x2+s2


    updated_C1=[x1/len(X1),x2/len(X1)]
    x1=0.0
    x2=0.0
    s1=0.0
    s2=0.0
    for j in np.arange(0,len(X2)):
        s1=X2[j][0]
        x1=x1+s1
    for j in np.arange(0,len(X2)):
        s2=X2[j][1]
        x2=x2+s2


    updated_C2=[x1/len(X2),x2/len(X2)]
    shift1=math.sqrt(math.pow(C1[0]-updated_C1[0],2)+math.pow(C1[1]-updated_C1[1],2))
    shift2=math.sqrt(math.pow(C2[0]-updated_C2[0],2)+math.pow(C2[1]-updated_C2[1],2))
    itr=itr+1
    C1=updated_C1
    C2=updated_C2
    X1_arr = np.array(X1)
    X2_arr = np.array(X2)
    plt.figure(figsize=(8, 5))
    plt.scatter(X1_arr[:,0], X1_arr[:,1], color='crimson', label='Cluster 1', alpha=0.7, edgecolors='k',s=35)
    plt.scatter(X2_arr[:,0], X2_arr[:,1], color='limegreen', label='Cluster 2', alpha=0.7, edgecolors='k', s=35)

    plt.scatter(C1[0],C1[1], color='black', alpha=1)
    plt.scatter(C2[0],C2[1], color='yellow', alpha=1)
    plt.show()

            
```
